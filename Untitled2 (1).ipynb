```python
def explain_mri_variable(col):
    if col == "NACCID":
        return "شناسه یکتای شرکت‌کننده", "شناسه"

    if col == "NACCUDSD":
        return "وضعیت تشخیصی شناختی شرکت‌کننده", "تشخیصی و بالینی"

    hemisphere = "نیم‌کره چپ" if col.startswith("LH_") else "نیم‌کره راست"

    name = col.replace("LH_", "").replace("RH_", "")

    if name.endswith("_GVOL"):
        region = name.replace("_GVOL", "")
        measurement = "حجم ماده خاکستری"
        category = "تصویربرداری MRI"
    elif name.endswith("_AVGTH"):
        region = name.replace("_AVGTH", "")
        measurement = "میانگین ضخامت قشر مغز"
        category = "تصویربرداری MRI"
    else:
        return "نیازمند بررسی", "سایر"

    regions_fa = {
        "CAUDALANTERIORCINGULATE": "بخش خلفی ناحیه کمربندی قدامی",
        "CAUDALMIDDLEFRONTAL": "بخش خلفی قشر پیشانی میانی",
        "CUNEUS": "کونئوس",
        "ENTORHINAL": "ناحیه انتورینال",
        "FUSIFORM": "شکنج دوکی‌شکل",
        "INFERIORPARIETAL": "ناحیه آهیانه‌ای تحتانی",
        "INFERIORTEMPORAL": "ناحیه گیجگاهی تحتانی",
        "ISTHMUSCINGULATE": "بخش باریک ناحیه کمربندی",
        "LATERALOCCIPITAL": "ناحیه جانبی لوب پس‌سری",
        "LATERALORBITOFRONTAL": "قشر پیشانی حدقه‌ای جانبی",
        "LINGUAL": "شکنج زبانی",
        "MEDIALORBITOFRONTAL": "قشر پیشانی حدقه‌ای میانی",
        "MIDDLETEMPORAL": "شکنج گیجگاهی میانی",
        "PARAHIPPOCAMPAL": "ناحیه پاراهیپوکامپال",
        "PARACENTRAL": "ناحیه پاراسنترال",
        "PARSOPERCULARIS": "بخش پوششی شکنج پیشانی تحتانی",
        "PARSORBITALIS": "بخش حدقه‌ای شکنج پیشانی تحتانی",
        "PARSTRIANGULARIS": "بخش مثلثی شکنج پیشانی تحتانی",
        "PERICALCARINE": "ناحیه پیرامون شیار کالکارین",
        "POSTCENTRAL": "شکنج پس‌مرکزی",
        "POSTERIORCINGULATE": "ناحیه کمربندی خلفی",
        "PRECENTRAL": "شکنج پیش‌مرکزی",
        "PRECUNEUS": "ناحیه پیش‌کونئوس",
        "ROSTRALANTERIORCINGULATE": "بخش جلویی ناحیه کمربندی قدامی",
        "ROSTRALMIDDLEFRONTAL": "بخش جلویی قشر پیشانی میانی",
        "SUPERIORFRONTAL": "قشر پیشانی فوقانی",
        "SUPERIORPARIETAL": "ناحیه آهیانه‌ای فوقانی",
        "SUPERIORTEMPORAL": "ناحیه گیجگاهی فوقانی",
        "SUPRAMARGINAL": "شکنج فوق‌حاشیه‌ای",
        "TRANSVERSETEMPORAL": "شکنج گیجگاهی عرضی",
        "INSULA": "قشر جزیره‌ای"
    }

    region_fa = regions_fa.get(region, region)
    description = f"{measurement} در {region_fa} در {hemisphere}"

    return description, category


rows = []

for i, col in enumerate(mri.columns, start=1):
    description, category = explain_mri_variable(col)

    rows.append({
        "شماره": i,
        "نام متغیر": col,
        "نوع اندازه گیری": description,
        "دسته‌بندی": category
    })

mri_dictionary = pd.DataFrame(rows)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

display(mri_dictionary)
```

```python
quantitative_vars = {
    "UDS": {
        "data": uds,
        "variables": ["NACCAGE", "MOCATOTS"]
    },
    "MRI": {
        "data": mri,
        "variables": [
            "LH_ENTORHINAL_GVOL",
            "RH_ENTORHINAL_GVOL",
            "LH_PARAHIPPOCAMPAL_GVOL",
            "RH_PARAHIPPOCAMPAL_GVOL"
        ]
    }
}

results = []

for file_name, info in quantitative_vars.items():
    df = info["data"]

    for var in info["variables"]:
        x = df[var].dropna()

        q1 = x.quantile(0.25)
        q2 = x.quantile(0.50)
        q3 = x.quantile(0.75)

        results.append({
            "File": file_name,
            "Variable": var,
            "N": len(x),
            "Mean": x.mean(),
            "Median": x.median(),
            "Standard Deviation": x.std(),
            "Minimum": x.min(),
            "Q1 (25%)": q1,
            "Q2 (50%)": q2,
            "Q3 (75%)": q3,
            "Maximum": x.max(),
            "IQR": q3 - q1
        })

summary_table = pd.DataFrame(results)

display(summary_table.round(3))
```
