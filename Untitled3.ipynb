```python
import pandas as pd

uds = pd.read_csv("udslong_imputed.csv")
mri = pd.read_csv("mriscanclean.csv")

datasets = {
    "Uds": uds,
    "MRI": mri
}
```

```python
data = pd.merge(
    uds,
    mri,
    on="NACCID",
    how="inner",
    validate="one_to_one"
)

variables = [
    "NACCID",
    "MOCATOTS",
    "LH_ENTORHINAL_GVOL",
    "RH_ENTORHINAL_GVOL",
    "LH_PARAHIPPOCAMPAL_GVOL",
    "RH_PARAHIPPOCAMPAL_GVOL"
]

df = data[variables].dropna()

Q1_cognitive = df["MOCATOTS"].quantile(0.25)
A = df["MOCATOTS"] < Q1_cognitive

mri_variables = [
    "LH_ENTORHINAL_GVOL",
    "RH_ENTORHINAL_GVOL",
    "LH_PARAHIPPOCAMPAL_GVOL",
    "RH_PARAHIPPOCAMPAL_GVOL"
]

n = len(df)

print("Number of participants:", n)
print("Q1 of MOCATOTS:", Q1_cognitive)
print("P(A):", A.sum() / n)

for variable in mri_variables:

    Q1 = df[variable].quantile(0.25)

    B = df[variable] < Q1

    P_B = B.sum() / n

    P_AB = (A & B).sum() / n

    P_A_given_B = (A & B).sum() / B.sum()

    print("\n------------------------------")
    print("MRI variable:", variable)
    print("Q1:", Q1)

    print("Count of B:", B.sum())
    print("P(B):", P_B)

    print("Count of A and B:", (A & B).sum())
    print("P(A and B):", P_AB)

    print("P(A | B):", P_A_given_B)
    print("P(A | B) percentage:", P_A_given_B * 100)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import probplot, norm

x = mri["LH_ENTORHINAL_GVOL"].dropna()

mean = x.mean()
variance = x.var()


Q1 = x.quantile(0.25)
Q3 = x.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = x[(x < lower_bound) | (x > upper_bound)]

print("Variable: LH_ENTORHINAL_GVOL")
print("Mean:", mean)
print("Variance:", variance)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

print("Number of outliers:", len(outliers))
print("Outlier values:")
print(outliers.sort_values().to_list())

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].hist(
    x,
    bins=200,
    density=True,
    alpha=0.6,
    edgecolor="black",
    label="Observed data"
)

axes[0].set_xlabel("LH_ENTORHINAL_GVOL")
axes[0].set_ylabel("Density")
axes[0].set_title("Histogram and Normal Curve")
axes[0].legend()
axes[0].grid(alpha=0.2)

probplot(x, dist="norm", plot=axes[1])

axes[1].set_title("Q-Q Plot")
axes[1].set_xlabel("Theoretical Normal Quantiles")
axes[1].set_ylabel("Observed Sample Quantiles")
axes[1].grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

uds_selected = uds[["NACCID", "MOCATOTS"]].copy()
mri_selected = mri[["NACCID", "LH_ENTORHINAL_GVOL"]].copy()

data = pd.merge(
    uds_selected,
    mri_selected,
    on="NACCID",
    how="inner"
)


covariance = data["MOCATOTS"].cov(
    data["LH_ENTORHINAL_GVOL"]
)
slope, intercept = np.polyfit(
    data["LH_ENTORHINAL_GVOL"],
    data["MOCATOTS"],
    1
)
line_x = np.linspace(
    data["LH_ENTORHINAL_GVOL"].min(),
    data["LH_ENTORHINAL_GVOL"].max(),
    100
)

plt.figure(figsize=(8, 6))

plt.scatter(
    data["LH_ENTORHINAL_GVOL"],
    data["MOCATOTS"],
    alpha=0.5
)

plt.plot(
    line_x,
    slope * line_x + intercept,
    color="red",
    linewidth=2,
    label="Trend line"
)

plt.xlabel("LH_ENTORHINAL_GVOL")
plt.ylabel("MOCATOTS")
plt.title("Cognitive Score vs Entorhinal Gray Matter Volume")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Covariance:", covariance)
```

```python
A = uds["MOCATOTS"] < 23
B = mri["LH_ENTORHINAL_GVOL"] < 1295

P_A = A.sum() / len(A)
P_B = B.sum() / len(B)
P_A_given_B = (A & B).sum() / B.sum()

P_B_given_A = (P_A_given_B * P_B) / P_A

print("Bayes' Theorem:")
print("P(B|A) = P(A|B) * P(B) / P(A)")

print("\nSubstitution:")
print(f"P(B|A) = ({P_A_given_B:.4f} * {P_B:.4f}) / {P_A:.4f}")

print("\nFinal Result:")
print(f"P(B|A) = {P_B_given_A:.4f}")
```
